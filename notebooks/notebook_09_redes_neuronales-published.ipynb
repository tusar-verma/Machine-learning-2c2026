````python
class Capa:
    def __init__(self, neuronas):
        self.neuronas = neuronas

    def forward(self, inputs, weights, bias, activation):
        """
        Forward Propagation de la capa
        """
        z =  np.dot(inputs, weights.T) + bias

        if activation == 'relu':
            a = self.relu(z)
        elif activation == 'softmax':
            a = self.softmax(z)

        return a, z

    def relu(self, inputs):
        """
        ReLU: función de activación
        """
        return np.maximum(0, inputs)

    def softmax(self, inputs):
        """
        Softmax: función de activación
        """
        exp_scores = np.exp(inputs)
        probs = exp_scores / np.sum(exp_scores, axis=1, keepdims=True)
        return probs

    def backward(self, dA_curr, W_curr, Z_curr, A_prev, activation):
        """
        Backward Propagation de la capa
        """
        # si la activación de la capa siguiente (estamos yendo backwards) es ReLU, se recibirá en dA_curr la derivada de la perdida con respecto a la salida de la capa actual
        # en cambio si es softmax, se recibirá directamente la derivada de la perdida con respecto a la entrada de softmax (la combinación lineal de los pesos). Esto por una simplificación en
        # el calculo de las derivadas cuando la función de perdida es cross-entropy.

        if activation == 'softmax':
            dW = np.dot(A_prev.T, dA_curr)
            # producto interno con vector de 1 es equivalente a sumar todos los elementos
            # la derivada de Aprev W + b con respecto a b es un vector de 1 con dimensión de b
            db = np.sum(dA_curr, axis=0, keepdims=True)
            dA = np.dot(dA_curr, W_curr)
        else:
            dZ = self.relu_derivative(dA_curr, Z_curr)
            dW = np.dot(A_prev.T, dZ)
            db = np.sum(dZ, axis=0, keepdims=True)
            dA = np.dot(dZ, W_curr)

        return dA, dW, db

    def relu_derivative(self, dA, Z):
        """
        ReLU: gradiente de ReLU (derivada de la perdida con respecto a ReLU)
        """
        dZ = np.array(dA, copy = True)
        # la derivada de relu es 1 si Z>0 y 0 si Z<=0.
        dZ[Z <= 0] = 0
        return dZ
````

````python
from numpy._core.multiarray import arange
class RedNeuronal:
    def __init__(self, learning_rate=0.01):
        self.red = [] ## capas
        self.arquitectura = [] ## mapeo de entradas -> salidas
        self.pesos = [] ## W, b
        self.memoria = [] ## Z, A
        self.gradientes = [] ## dW, db
        self.lr = learning_rate

    def add(self, capa):
        """
        Agregar capa a la red
        """
        self.red.append(capa)




    def _compile(self, data):
        """
        Inicializar la arquitectura
        """
        for i, capa in enumerate(self.red):
          if i == 0:
            # la primer capa recibe de entrada la cantidad de features de los datos y tiene de salida su cantidad de neuronas
            self.arquitectura.append({'input_dim': data.shape[1],
                                    'output_dim': self.red[i].neuronas,
                                    'activation':'relu'})
          elif i > 0 and i < len(self.red)-1:
            # las capas ocultas reciben de entrada la cantidad de neuronas de la capa anterior y de salida su cantidad de neuronas
            self.arquitectura.append({'input_dim': self.red[i-1].neuronas,
                                    'output_dim': self.red[i].neuronas,
                                    'activation':'relu'})
          else:
            # la ultima misma logica para entrada y salida pero la activación es softmax
            self.arquitectura.append({'input_dim': self.red[i-1].neuronas,
                                    'output_dim': self.red[i].neuronas,
                                    'activation':'softmax'})




    def print_dim_pesos(self):
      for i in range(len(self.pesos)):
        print(f"capa {i}: w={self.pesos[i]['W'].shape} - b={self.pesos[i]['b'].shape}")

    def _init_weights(self, data):
        """
        Inicializar arquitectura y los pesos
        """
        self._compile(data)

        np.random.seed(67)

        # el rango de sampleo del uniforme y el valor inicial de b se lo copie a la resolución de más abajo
        for i in range(len(self.arquitectura)):
          self.pesos.append({
              "W": np.random.uniform(low=-1, high=1, size=(self.arquitectura[i]['input_dim'], self.arquitectura[i]['output_dim'])),
              "b": np.zeros((1, self.arquitectura[i]['output_dim']))
          })

        return self



    def _forwardprop(self, data):
        """
        Pasada forward completa por la red
        """
        # Pasos:
        # 1: recorrer cada capa haciendo la cuenta W_actual A_anterior + b_actual = Z_actual
        #   1.1: guardarse Z_actual para luego usarlo en backprop
        #   1.2: buscar en la arquitectura la activación a usar en la capa actual, calcular A_actual y guardarla
        # 2. devolver el resultado final de la ultima capa

        A_curr = data

        for i in range(len(self.pesos)):
            A_prev = A_curr
            A_curr, Z_curr = self.red[i].forward(A_prev, self.pesos[i]['W'].T, self.pesos[i]['b'], self.arquitectura[i]['activation'])
            self.memoria.append({'A': A_curr, 'Z': Z_curr})

        return A_curr

    def _backprop(self, predicted, actual):
        """
        Pasada backward completa por la red
        """
        # Esta la copie.

        # Se derivada de la función de perdida con respecto a la función de activación softmax para la última capa de la red
        # (Softmax + Categorical Cross-Entropy)
        # actual tiene las clases verdaderas
        # predicted las predichas por la red (el output de la última capa).
        num_samples = len(actual)

        ## compute the gradient on predictions

        ## aca se calcula \hat(Y) - Y
        dscores = predicted
        dscores[range(num_samples),actual] -= 1
        dscores /= num_samples

        dA_prev = dscores

        # Luego se itera por cada capa llamando al método backward pasandole:
        #   la derivada de la función de pérdida con respecto a la salida de la capa actual,
        #   los pesos de la capa actual,
        #   el producto Z_act = W_actual A_ant + b_actual,
        #   la entrada de la capa actual (A_prev) y la función de activación de la capa actual.

        for idx, layer in reversed(list(enumerate(self.red))):
            dA_curr = dA_prev

            A_prev = self.memoria[idx]['A']
            Z_curr = self.memoria[idx]['Z']
            W_curr = self.pesos[idx]['W']

            activation = self.arquitectura[idx]['activation']

            dA_prev, dW_curr, db_curr = layer.backward(dA_curr, W_curr.T, Z_curr, A_prev, activation)

            self.gradientes.append({'dW':dW_curr, 'db':db_curr})

        self.gradientes = list(reversed(self.gradientes))  # Reverse the gradients list


    def _update(self):
        """
        Actualizar el modelo --> lr * gradiente
        """
        # se actualiza los pesos de todas las capas según los gradientes calculados por backprop

        for idx, layer in enumerate(self.red):
            self.pesos[idx]['W'] -= self.lr * self.gradientes[idx]['dW']
            self.pesos[idx]['b'] -= self.lr * self.gradientes[idx]['db']

    def _get_accuracy(self, predicted, actual):
        """
        Calcular accuracy después de cada iteración
        """
        # Para cada instancia, tomamos el vector de probabilidades por clases predichas y nos quedamos con la clase de mayor probabilidad. Calculamos acurracy sobre esa predicción.
        return np.mean(np.argmax(predicted, axis=1)==actual)

    def _calculate_loss(self, predicted, actual):
        """
        Calcular cross-entropy loss después de cada iteración
        """

        # fancy indexing. Sirve para extraer un elemento específico por cada fila sin usar un bucle
        # en predicted se tiene un elemento por instancia. Cada elemento es a su vez una lista de probas de cada clase
        # range(len(actual)) devuelve [0,1,2,...] una lista para indexar a cada instancia
        # actual es una lista con la clase real de cada instancia [1, 0, 2, 3, ...] por ejemplo.
        # predicted[range(len(actual)),actual] devuelve una lista de la prob asignada a la clase real de cada instancia
        # a esto ultimo se calcula el log y se promedia entre todas las instancias.
        return -np.log(predicted[range(len(actual)),actual]).mean()

    def train(self, X_train, y_train, epochs):
        """
        Entrenar el modelo Stochastic Gradient Descent
        """
        self.loss = []
        self.accuracy = []

        self._init_weights(X_train)

        for i in range(epochs):
            yhat = self._forwardprop(X_train)
            self.accuracy.append(self._get_accuracy(predicted=yhat, actual=y_train))
            self.loss.append(self._calculate_loss(predicted=yhat, actual=y_train))

            self._backprop(predicted=yhat, actual=y_train)

            self._update()

            if i % 20 == 0:
                s = 'EPOCH: {}, ACCURACY: {}, LOSS: {}'.format(i, self.accuracy[-1], self.loss[-1])
                print(s)

        return (self.accuracy, self.loss)
````

````python
# graficos accuracy y loss por epoch utilizando las implementaciones de arriba
import matplotlib.pyplot as plt
import numpy as np

# Usamos las clases Capa y RedNeuronal ya declaradas arriba en el notebook
primer_capa = Capa(6)
segunda_capa = Capa(8)
tercer_capa = Capa(10)
cuarta_capa = Capa(3)

nn = RedNeuronal(learning_rate=0.01)
nn.add(primer_capa)
nn.add(segunda_capa)
nn.add(tercer_capa)
nn.add(cuarta_capa)

# Entrenamos la red por 200 épocas
accuracy_history, loss_history = nn.train(X, y, 1000)
epochs_range = range(len(accuracy_history))

# Graficamos la pérdida y el accuracy
plt.figure(figsize=(12, 5))

# Gráfico de Loss
plt.subplot(1, 2, 1)
plt.plot(epochs_range, loss_history, label='Loss', color='red')
plt.title('Pérdida (Loss) por Epoch')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.grid(True)
plt.legend()

# Gráfico de Accuracy
plt.subplot(1, 2, 2)
plt.plot(epochs_range, accuracy_history, label='Accuracy', color='blue')
plt.title('Precisión (Accuracy) por Epoch')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
````

````python
class Capa:
    def __init__(self, neuronas):
        self.neuronas = neuronas

    def forward(self, inputs, weights, bias, activation):
        """
        Forward Propagation de la capa
        """
        Z_curr = np.dot(inputs, weights.T) + bias

        if activation == 'relu':
            A_curr = self.relu(inputs=Z_curr)
        elif activation == 'softmax':
            A_curr = self.softmax(inputs=Z_curr)

        return A_curr, Z_curr

    def relu(self, inputs):
        """
        ReLU: función de activación
        """

        return np.maximum(0, inputs)

    def softmax(self, inputs):
        """
        Softmax: función de activación
        """
        exp_scores = np.exp(inputs)
        probs = exp_scores / np.sum(exp_scores, axis=1, keepdims=True)
        return probs

    def backward(self, dA_curr, W_curr, Z_curr, A_prev, activation):
        """
        Backward Propagation de la capa
        """
        if activation == 'softmax':
            dW = np.dot(A_prev.T, dA_curr)
            db = np.sum(dA_curr, axis=0, keepdims=True)
            dA = np.dot(dA_curr, W_curr)
        else:
            dZ = self.relu_derivative(dA_curr, Z_curr)
            dW = np.dot(A_prev.T, dZ)
            db = np.sum(dZ, axis=0, keepdims=True)
            dA = np.dot(dZ, W_curr)

        return dA, dW, db

    def relu_derivative(self, dA, Z):
        """
        ReLU: gradiente de ReLU
        """
        dZ = np.array(dA, copy = True)
        dZ[Z <= 0] = 0
        return dZ
````

````python
class RedNeuronal:
    def __init__(self, learning_rate=0.01):
        self.red = [] ## capas
        self.arquitectura = [] ## mapeo de entradas -> salidas
        self.pesos = [] ## W, b
        self.memoria = [] ## Z, A
        self.gradientes = [] ## dW, db
        self.lr = learning_rate

    def add(self, capa):
        """
        Agregar capa a la red
        """
        self.red.append(capa)

    def _compile(self, data):
        """
        Inicializar la arquitectura
        """
        for idx, _ in enumerate(self.red):
            if idx == 0:
                self.arquitectura.append({'input_dim': data.shape[1],
                                        'output_dim': self.red[idx].neuronas,
                                        'activation':'relu'})
            elif idx > 0 and idx < len(self.red)-1:
                self.arquitectura.append({'input_dim': self.red[idx-1].neuronas,
                                        'output_dim': self.red[idx].neuronas,
                                        'activation':'relu'})
            else:
                self.arquitectura.append({'input_dim': self.red[idx-1].neuronas,
                                        'output_dim': self.red[idx].neuronas,
                                        'activation':'softmax'})
        return self

    def _init_weights(self, data):
        """
        Inicializar arquitectura y los pesos
        """
        self._compile(data)

        np.random.seed(99)

        for i in range(len(self.arquitectura)):
            self.pesos.append({
                'W':np.random.uniform(low=-1, high=1,
                        size=(self.arquitectura[i]['input_dim'],
                            self.arquitectura[i]['output_dim']
                            )),
                'b':np.zeros((1, self.arquitectura[i]['output_dim']))})

        return self

    def _forwardprop(self, data):
        """
        Pasada forward completa por la red
        """
        A_curr = data

        for i in range(len(self.pesos)):
            A_prev = A_curr
            A_curr, Z_curr = self.red[i].forward(inputs=A_prev,
                                                    weights=self.pesos[i]['W'].T,
                                                    bias=self.pesos[i]['b'],
                                                    activation=self.arquitectura[i]['activation'])

            self.memoria.append({'inputs':A_prev, 'Z':Z_curr})

        return A_curr

    def _backprop(self, predicted, actual):
        """
        Pasada backward completa por la red
        """
        num_samples = len(actual)

        ## compute the gradient on predictions
        dscores = predicted
        dscores[range(num_samples),actual] -= 1
        dscores /= num_samples

        dA_prev = dscores

        for idx, layer in reversed(list(enumerate(self.red))):
            dA_curr = dA_prev

            A_prev = self.memoria[idx]['inputs']
            Z_curr = self.memoria[idx]['Z']
            W_curr = self.pesos[idx]['W']

            activation = self.arquitectura[idx]['activation']

            dA_prev, dW_curr, db_curr = layer.backward(dA_curr, W_curr.T, Z_curr, A_prev, activation)

            self.gradientes.append({'dW':dW_curr, 'db':db_curr})

        self.gradientes = list(reversed(self.gradientes))  # Reverse the gradients list

    def _update(self):
        """
        Actualizar el modelo --> lr * gradiente
        """
        lr = self.lr
        for idx, layer in enumerate(self.red):
            self.pesos[idx]['W'] -= lr * self.gradientes[idx]['dW']
            self.pesos[idx]['b'] -= lr * self.gradientes[idx]['db']

    def _get_accuracy(self, predicted, actual):
        """
        Calcular accuracy después de cada iteración
        """
        return np.mean(np.argmax(predicted, axis=1)==actual)

    def _calculate_loss(self, predicted, actual):
        """
        Calculate cross-entropy loss after each iteration
        """
        samples = len(actual)

        correct_logprobs = -np.log(predicted[range(samples),actual])
        data_loss = np.sum(correct_logprobs)/samples

        return data_loss

    def train(self, X_train, y_train, epochs):
        """
        Entrenar el modelo Stochastic Gradient Descent
        """
        self.loss = []
        self.accuracy = []

        self._init_weights(X_train)

        for i in range(epochs):
            yhat = self._forwardprop(X_train)
            self.accuracy.append(self._get_accuracy(predicted=yhat, actual=y_train))
            self.loss.append(self._calculate_loss(predicted=yhat, actual=y_train))

            self._backprop(predicted=yhat, actual=y_train)

            self._update()

            if i % 20 == 0:
                s = 'EPOCH: {}, ACCURACY: {}, LOSS: {}'.format(i, self.accuracy[-1], self.loss[-1])
                print(s)

        return (self.accuracy, self.loss)
````
